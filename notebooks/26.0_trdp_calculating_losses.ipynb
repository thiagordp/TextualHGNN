## Entropy loss

$$\mathcal{L}_{\text{entropy}} = -\sum_{i=1}^N \sum_{j=1}^C S_{ij} \log S_{ij}$$


```python
from matplotlib import pyplot as plt
import random
import itertools
import torch

grid = [0.0, 0.25, 0.5, 0.6, 0.75, 0.9, 1.0]

combinations = list(itertools.product(grid, grid, grid, grid))

entropies= []
for n1, n2, n3, n4 in combinations[:50]:

    # if random.random() < 0.1:
    #     continue

    print("-" * 50)
    print(n1, n2, n3, n4)
    t = torch.tensor([n1, n2, n3, n4])
    t = torch.softmax(t, dim=0)
    print(t)
    entropy = torch.sum(-t * torch.log(t), dim=0)
    print(entropy)
    entropies.append(entropy)

plt.hist(entropies)
```
