```python
from src.multi_graph.data_preprocessing import preprocessing_legal_pt, preprocess_text

#!/usr/bin/env python3
"""
Dependency Parsing Tree Visualizer using spaCy

This script processes a string document and visualizes its dependency
parsing tree using spaCy's displaCy. The output is saved as an HTML file.

Usage:
    python dependency_tree.py "Your sentence here."
"""

import sys
import spacy
from spacy import displacy
import os


def load_model(model_name: str = "en_core_web_lg"):
    """
    Load a spaCy language model with error handling.
    """
    try:
        return spacy.load(model_name)
    except OSError:
        print(f"Model '{model_name}' not found. Attempting to download...")
        from spacy.cli import download
        download(model_name)
        return spacy.load(model_name)


def visualize_dependency_tree(text: str, nlp, output_file: str = "dependency_tree.html" ):
    """
    Process the text and save the dependency parsing tree as an HTML file.

    Args:
        text (str): Input document string.
        nlp (spacy.language.Language): spaCy language model.
        output_file (str): Output HTML file name.
    """
    if not text.strip():
        raise ValueError("Input text is empty. Please provide valid text.")

    doc = nlp(text)

    # Render dependency tree
    html = displacy.render(doc, style="dep", page=True)

    # Save to file
    with open(output_file, "w", encoding="utf-8") as f:
        f.write(html)

    print(f"Dependency parsing tree saved to '{output_file}'.")


def main():
    text = """
"This film, which many critics have praised, is, in the final analysis, not really good."
    """.strip()
    nlp = load_model(model_name="en_core_web_lg")
    # text = preprocessing_legal_pt(text, nlp)
    text = preprocess_text(text, nlp)

    print(f"Text: {text}")
    try:
        visualize_dependency_tree(text.strip(), nlp=nlp)
    except Exception as e:
        print(f"Error: {e}")


if __name__ == "__main__":
    main()
```
