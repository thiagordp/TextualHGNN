# Disambiguition with wordnet.

```python
import nltk
from nltk.wsd import lesk
from nltk.corpus import wordnet

# --- Setup: Download necessary NLTK data ---
# This only needs to be run once.
try:
    nltk.data.find('corpora/wordnet.zip')
except nltk.downloader.DownloadError:
    print("--- Downloading 'wordnet' from NLTK (one-time setup) ---")
    nltk.download('wordnet', quiet=True)
try:
    nltk.data.find('tokenizers/punkt.zip')
except nltk.downloader.DownloadError:
    print("--- Downloading 'punkt' from NLTK (one-time setup) ---")
    nltk.download('punkt', quiet=True)
print("NLTK resources are ready.\n")


def run_wsd_test():
    """
    Demonstrates that a Word Sense Disambiguation algorithm can identify the
    correct 'sense' of an ambiguous word based on its context.
    """
    # --- Define Sentences with an Ambiguous Word ---
    # "bank" has a financial meaning here
    sentence_1 = "I went to the bank to deposit my money."
    # "bank" has a geographical meaning here
    sentence_2 = "We had a picnic on the river bank."

    ambiguous_word = "bank"

    print("--- Performing Word Sense Disambiguation ---")

    # 1. Disambiguate the word in the first sentence
    # We provide the full sentence as context and specify the target word
    print(f"Analyzing Sentence 1: '{sentence_1}'")
    sense_1 = lesk(sentence_1.split(), ambiguous_word, 'n') # 'n' specifies to look for noun senses

    if sense_1:
        print(f"✅ Identified Sense: {sense_1.name()}")
        print(f"   Definition: {sense_1.definition()}")
    else:
        print("❌ Could not disambiguate 'bank' in the first sentence.")

    print("-" * 20)

    # 2. Disambiguate the word in the second sentence
    print(f"Analyzing Sentence 2: '{sentence_2}'")
    sense_2 = lesk(sentence_2.split(), ambiguous_word, 'n')

    if sense_2:
        print(f"✅ Identified Sense: {sense_2.name()}")
        print(f"   Definition: {sense_2.definition()}")
    else:
        print("❌ Could not disambiguate 'bank' in the second sentence.")

    print("\n--- Analysis ---")
    if sense_1 and sense_2 and sense_1 != sense_2:
        print("✅ Result: The algorithm successfully identified two different synsets for 'bank'.")
        print("This proves that an explicit WSD step can resolve ambiguity before graph construction.")
        print(f"The graph builder would create distinct nodes for '{sense_1.name()}' and '{sense_2.name()}'.")
    else:
        print("❌ Result: The algorithm failed to distinguish between the two senses.")

if __name__ == '__main__':
    run_wsd_test()
```

### Portuguese


```python
import nltk
from nltk.wsd import lesk
from nltk.corpus import wordnet

# --- Setup: Download necessary NLTK data for Portuguese ---
# This only needs to be run once.
try:
    # WordNet is the base corpus
    nltk.data.find('corpora/wordnet.zip')
except nltk.downloader.DownloadError:
    print("--- Downloading 'wordnet' from NLTK (one-time setup) ---")
    nltk.download('wordnet', quiet=True)

try:
    # Open Multilingual Wordnet (OMW) contains Portuguese data
    nltk.data.find('corpora/omw-1.4.zip')
except nltk.downloader.DownloadError:
    print("--- Downloading 'omw-1.4' for multilingual support (one-time setup) ---")
    nltk.download('omw-1.4', quiet=True)

try:
    # Punkt is needed for tokenization
    nltk.data.find('tokenizers/punkt.zip')
except nltk.downloader.DownloadError:
    print("--- Downloading 'punkt' from NLTK (one-time setup) ---")
    nltk.download('punkt', quiet=True)
print("NLTK resources are ready.\n")


def run_wsd_test_portuguese():
    """
    Demonstrates that a Word Sense Disambiguation algorithm can identify the
    correct 'sense' of an ambiguous word in Portuguese.
    """
    # --- Define Sentences with an Ambiguous Word ---
    # "banco" has a financial meaning here
    sentence_1 = "Eu fui ao banco para depositar dinheiro."
    # "banco" has a furniture meaning here
    sentence_2 = "Nós sentamos no banco da praça para conversar."

    ambiguous_word = "banco"

    print("--- Performing Word Sense Disambiguation in Portuguese ---")

    # 1. Disambiguate the word in the first sentence
    print(f"Analyzing Sentence 1: '{sentence_1}'")
    # --- The key change is lang='por' ---
    sense_1 = lesk(nltk.word_tokenize(sentence_1, language='portuguese'), ambiguous_word, 'n', lang='por')

    if sense_1:
        print(f"✅ Identified Sense: {sense_1.name()}")
        # Display definitions in both Portuguese (if available) and English
        print(f"   Definition (PT): {sense_1.definition()}")
        print(f"   Definition (EN): {wordnet.synset(sense_1.name()).definition()}")
    else:
        print("❌ Could not disambiguate 'banco' in the first sentence.")

    print("-" * 20)

    # 2. Disambiguate the word in the second sentence
    print(f"Analyzing Sentence 2: '{sentence_2}'")
    sense_2 = lesk(nltk.word_tokenize(sentence_2, language='portuguese'), ambiguous_word, 'n', lang='por')

    if sense_2:
        print(f"✅ Identified Sense: {sense_2.name()}")
        print(f"   Definition (PT): {sense_2.definition()}")
        print(f"   Definition (EN): {wordnet.synset(sense_2.name()).definition()}")
    else:
        print("❌ Could not disambiguate 'banco' in the second sentence.")

    print("\n--- Analysis ---")
    if sense_1 and sense_2 and sense_1 != sense_2:
        print("✅ Result: The algorithm successfully identified two different synsets for 'banco'.")
        print("This proves that an explicit WSD step can resolve ambiguity in Portuguese before graph construction.")
    else:
        print("❌ Result: The algorithm failed to distinguish between the two senses.")

if __name__ == '__main__':
    run_wsd_test_portuguese()
```

### Transformer based WSD

```python
import torch
from transformers import AutoTokenizer, AutoModel
from nltk.corpus import wordnet as wn
from torch.nn.functional import cosine_similarity
import nltk

# --- Setup: Download necessary NLTK data ---
try:
    nltk.data.find('corpora/wordnet.zip')
    nltk.data.find('corpora/omw-1.4.zip')
except nltk.downloader.DownloadError:
    print("--- Downloading NLTK resources (wordnet, omw-1.4) ---")
    nltk.download('wordnet', quiet=True)
    nltk.download('omw-1.4', quiet=True)
print("NLTK resources are ready.\n")

def get_sentence_embedding(text, model, tokenizer):
    """Generates a single embedding vector for a full sentence."""
    inputs = tokenizer(text, return_tensors='pt', truncation=True, max_length=512)
    with torch.no_grad():
        outputs = model(**inputs)
    # Use the embedding of the [CLS] token as the representation of the whole sentence
    return outputs.last_hidden_state[0, 0, :]

def run_transformer_wsd_test_portuguese():
    """
    Demonstrates robust Word Sense Disambiguation in Portuguese using a
    pre-trained Transformer model.
    """
    print("--- Loading Pre-trained Portuguese Model (BERTimbau) ---")
    model_name = 'neuralmind/bert-base-portuguese-cased'
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModel.from_pretrained(model_name)
    print("Model loaded successfully.\n")

    # --- Define Sentences with an Ambiguous Word ---
    sentence_1 = "Eu fui ao banco para depositar dinheiro."
    sentence_2 = "Nós sentamos no banco da praça para conversar."
    ambiguous_word = "banco"

    print("--- Performing Transformer-based WSD ---")

    # 1. Get all possible senses for "banco" in Portuguese
    banco_synsets = wn.synsets(ambiguous_word, lang='por')
    if not banco_synsets:
        print("Could not find synsets for 'banco' in Portuguese.")
        return

    # --- FIX: Correctly get definitions for the retrieved Portuguese synsets ---
    # For each Portuguese synset, look up its English definition for robustness with the embedding model.
    sense_definitions = []
    valid_synsets = []
    for syn in banco_synsets:
        try:
            # Use the synset name (e.g., 'bank.n.01') to get the English definition
            eng_syn = wn.synset(syn.name())
            sense_definitions.append(eng_syn.definition())
            valid_synsets.append(syn) # Keep track of the synsets that have valid definitions
        except:
            continue

    if not sense_definitions:
        print("Error: Could not retrieve any valid definitions for the synsets.")
        return

    sense_embeddings = torch.stack([get_sentence_embedding(d, model, tokenizer) for d in sense_definitions])

    # 3. Get the contextualized embedding of the first sentence
    context_embedding_1 = get_sentence_embedding(sentence_1, model, tokenizer)

    # 4. Find the best sense for sentence 1
    similarities_1 = cosine_similarity(context_embedding_1.unsqueeze(0), sense_embeddings)
    best_sense_idx_1 = torch.argmax(similarities_1)
    sense_1 = valid_synsets[best_sense_idx_1]

    print(f"Analyzing Sentence 1: '{sentence_1}'")
    print(f"✅ Best Match: {sense_1.name()} ({sense_1.definition()})")

    print("-" * 20)

    # 5. Get the contextualized embedding of the second sentence
    context_embedding_2 = get_sentence_embedding(sentence_2, model, tokenizer)

    # 6. Find the best sense for sentence 2
    similarities_2 = cosine_similarity(context_embedding_2.unsqueeze(0), sense_embeddings)
    best_sense_idx_2 = torch.argmax(similarities_2)
    sense_2 = valid_synsets[best_sense_idx_2]

    print(f"Analyzing Sentence 2: '{sentence_2}'")
    print(f"✅ Best Match: {sense_2.name()} ({sense_2.definition()})")

    print("\n--- Analysis ---")
    if sense_1 and sense_2 and sense_1 != sense_2:
        print("✅ Result: The Transformer-based method successfully identified two different synsets for 'banco'.")
        print("This approach correctly uses sentence-level semantic context to resolve ambiguity.")
    else:
        print("❌ Result: The method failed to distinguish between the two senses.")

if __name__ == '__main__':
    run_transformer_wsd_test_portuguese()
```

## Strategy 2 for Portuguese

```python
import torch
from transformers import AutoTokenizer, AutoModel
from torch.nn.functional import cosine_similarity
import warnings

# Suppress warnings for a cleaner output
warnings.filterwarnings("ignore", category=UserWarning)

def get_contextual_embedding_pt(text: str, target_word: str, model, tokenizer):
    """
    Finds a target word in a Portuguese text, tokenizes it, and returns the
    contextualized embedding for that specific word from a Transformer model.
    """
    # 1. Tokenize the input text
    inputs = tokenizer(text, return_tensors='pt')

    # 2. Get the model's output (the last hidden state)
    with torch.no_grad():
        outputs = model(**inputs)
        # Shape: [batch_size, num_tokens, embedding_dim]
        last_hidden_state = outputs.last_hidden_state.squeeze(0)

    # 3. Find the token(s) corresponding to our target word
    token_strings = tokenizer.convert_ids_to_tokens(inputs['input_ids'][0])

    # Handle casing by finding the lowercase version
    target_token_indices = [i for i, token in enumerate(token_strings) if token.lower() == target_word.lower()]

    if not target_token_indices:
        print(f"Warning: Target word '{target_word}' not found as a single token in '{text}'.")
        return None

    # For simplicity, we take the first occurrence.
    token_index = target_token_indices[0]

    # 4. Extract the embedding vector for the target token
    word_embedding = last_hidden_state[token_index]

    print(f"Text: '{text}'")
    print(f"Found '{target_word}' at token index: {token_index} (token: '{token_strings[token_index]}')")

    return word_embedding

def run_transformer_disambiguation_test_pt():
    """
    Demonstrates that a Portuguese Transformer model produces different
    embeddings for the same word used in different contexts.
    """
    print("--- Loading Pre-trained Portuguese Model (BERTimbau) ---")
    model_name = 'neuralmind/bert-base-portuguese-cased'
    tokenizer = AutoTokenizer.from_pretrained(model_name)
    model = AutoModel.from_pretrained(model_name)
    print("Model loaded successfully.\n")

    # --- Define Sentences with an Ambiguous Word ---
    sentence_1 = "Eu preciso depositar dinheiro no banco."
    sentence_2 = "Vamos sentar naquele banco do parque."
    target_word = "banco"

    print("--- Extracting Contextualized Embeddings ---")
    # Get the specific embedding for "banco" from the first sentence
    embedding_1 = get_contextual_embedding_pt(sentence_1, target_word, model, tokenizer)
    print("-" * 20)

    # Get the specific embedding for "banco" from the second sentence
    embedding_2 = get_contextual_embedding_pt(sentence_2, target_word, model, tokenizer)
    print("-" * 20)

    if embedding_1 is None or embedding_2 is None:
        print("\nCould not perform comparison.")
        return

    # --- Compare the Embeddings ---
    similarity = cosine_similarity(embedding_1.unsqueeze(0), embedding_2.unsqueeze(0)).item()

    print("\n--- Analysis ---")
    print(f"Cosine Similarity between the two vectors for the word 'banco': {similarity:.4f}")

    if similarity < 0.7:
        print("✅ Result: The similarity is LOW, as expected.")
        print("This proves the model generated two distinct, context-aware embeddings for the same word form.")
    else:
        print("❌ Result: The similarity is HIGH.")
        print("This would indicate the model did not effectively disambiguate the word based on context.")

if __name__ == '__main__':
    run_transformer_disambiguation_test_pt()
```
