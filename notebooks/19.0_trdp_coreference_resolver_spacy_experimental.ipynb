# CoreferenceResolver

```python
import spacy
from spacy.language import Language
from spacy_experimental.coref.coref_component import CoreferenceResolver

@Language.factory("coref")
def create_coref_component(nlp, name):
    return CoreferenceResolver(nlp.vocab, model=None, name=name)

# Carregar o modelo de linguagem em português
nlp = spacy.load("pt_core_news_lg")

# Adicionar o componente de resolução de co-referência ao pipeline
nlp.add_pipe("coref", last=True)

# Inicializar o pipeline
nlp.initialize(lambda: [])

# Texto de exemplo
texto = (
    "A Maria foi ao mercado. Depois, ela encontrou o João. "
    "Ele perguntou a ela se Maria queria ajuda com as compras."
)

# Processar o texto
doc = nlp(texto)

# Exibir as referências resolvidas
for key, spans in doc.spans.items():
    if key.startswith("coref_clusters"):
        print(key, [span.text for span in spans])
```
