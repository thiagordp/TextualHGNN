# STF Split

```python
import os
import shutil
import random
from collections import defaultdict

# CONFIGURATION
SOURCE_DIR = '../data/datasets/STF_HC/'    # Path to your source dataset
OUTPUT_DIR = '../data/datasets/STF_HC_split/'     # Path to save the split dataset
TRAIN_RATIO = 0.8         # 70% for training
VAL_RATIO = 0.10          # 15% for validation
TEST_RATIO = 0.10         # 15% for testing
SEED = 42                 # For reproducibility

random.seed(SEED)

# Helper Function to Create Folders
def create_dir(path):
    if not os.path.exists(path):
        os.makedirs(path)

# Get all files grouped by class
data = defaultdict(list)

for class_name in os.listdir(SOURCE_DIR):
    class_path = os.path.join(SOURCE_DIR, class_name)
    if os.path.isdir(class_path):
        for file in os.listdir(class_path):
            if file.endswith('.txt'):
                data[class_name].append(os.path.join(class_path, file))

# Perform Stratified Split
for split in ['train', 'validation', 'test']:
    for class_name in data.keys():
        create_dir(os.path.join(OUTPUT_DIR, split, class_name))

for class_name, files in data.items():
    random.shuffle(files)
    total = len(files)
    train_end = int(total * TRAIN_RATIO)
    val_end = train_end + int(total * VAL_RATIO)

    splits = {
        'train': files[:train_end],
        'validation': files[train_end:val_end],
        'test': files[val_end:]
    }

    for split, split_files in splits.items():
        for file_path in split_files:
            filename = os.path.basename(file_path)
            dest_path = os.path.join(OUTPUT_DIR, split, class_name, filename)
            shutil.copy(file_path, dest_path)

print("Dataset split completed successfully.")
```
