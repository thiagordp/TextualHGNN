```python
from sklearn.metrics import classification_report

# True labels
y_true = ['Preso'] * 151 + ['Solto'] * 52

# Dummy majority-class predictions (always predict 'Preso')
y_pred = ['Preso'] * len(y_true)

# Generate classification report
report = classification_report(y_true, y_pred, labels=['Preso', 'Solto'], zero_division=0, digits=3)
print(report)
```
