# Convert embeddings models from text to binary

```python
from gensim.models import KeyedVectors
```

```python
# Load the model in text format (slow, one-time operation)
model = KeyedVectors.load_word2vec_format("../data/external/embeddings/itwiki_20180420_100d.txt", binary=False)

# Save the model in binary format
model.save("../data/external/embeddings/itwiki_20180420_100d.bin")
```

```python
model = KeyedVectors.load("../data/external/embeddings/itwiki_20180420_100d.bin", mmap='r')
```

```python
model.most_similar(positive=["tutto"])
```
