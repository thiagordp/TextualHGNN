```python
import spacy

nlp = spacy.load("en_core_web_lg")

def compute_quality(text):
    """
    Returns a quality score (0-1) for a given text.
    High score = clean, readable text
    Low score = messy, noisy text
    """
    doc = nlp(text)
    total_tokens = len(doc)
    if total_tokens == 0:
        return 0.0

    # Good tokens
    good_tokens = sum(
        1 for t in doc if t.is_alpha or t.is_stop or t.is_punct or t.is_digit
    )
    # Score = good / total
    score = good_tokens / total_tokens

    return round(score, 3)

# Example usage
texts = [
    "Hello, this is a clean document! It has proper words and punctuation.",
    "Random chars: @$#%^&*!!! 😂😎",
    "1234567890"
]

for t in texts:
    print(f"Score: {compute_quality(t)} | Text: {t[:50]}...")
```
