# PMI

```python
import pickle

PMI_PATH = "../data/datasets/IMDB/test/interim/pmi_cache_phrase_subgraphs.pkl"
with open(PMI_PATH, 'rb') as f:
    pmi = pickle.load(f)
```

```python
pmi_sorted = sorted(pmi, key=pmi.get, reverse=True)
```

```python
print(f"values:\t{len(pmi_sorted)}")
```

```python
from matplotlib import pyplot as plt

for key in pmi_sorted[1000:2000]:
    print(f"Key: {key[1]} {key[0]}\tPMI: {pmi[key]:.3f}")
```

```python
values = [pmi[v] for v in pmi_sorted]
plt.hist(values, bins=20)
```

```python
import numpy as np

print("Mean: ", np.mean(values))
print("Standard Deviation: ", np.std(values))
print("Median: ", np.median(values))
print("Minimum: ", np.min(values))
print("Maximum: ", np.max(values))
print("Q1: ", np.quantile(values, 0.1))
print("Q3: ", np.quantile(values, 0.3))
print("Q5: ", np.quantile(values, 0.5))
print("Q7: ", np.quantile(values, 0.7))
print("Q9: ", np.quantile(values, 0.9))
print("Q10: ", np.quantile(values, 0.95))
```

```python
plt.figure(figsize=(16, 8), dpi=300)
plt.xticks(rotation=30, ha="right")
plt.grid(which="major", linestyle="-", linewidth="0.5")
plt.plot(values)
```
